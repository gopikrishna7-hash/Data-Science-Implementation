```python
from io import StringIO
import pandas as pd
data=""" col1,col2,col3,col4\n
1,2,3,4\n
num1,num2,num3\n
w,x,y,z """
stream=StringIO(data)
df=pd.read_csv(stream)
df.head()
```

```python
df.sort_values("Kills")
```

```python
q1=df["Damage_Dealt"].quantile(0.25)
q2=df["Damage_Dealt"].quantile(0.75)
iqr=q2-q1
lower=q1-1.5*iqr
upper=q2+1.5*iqr
print(df[(df["Damage_Dealt"] < lower) | (upper < df["Damage_Dealt"])])
```

```python
pd.pivot_table(
    df,
    values=["Damage_Dealt","Kills","Headshot_Percentage"],
    index="Tier",
    aggfunc="mean"
)
```

```python
df.to_csv("pubg_details.csv",index=False)
```
