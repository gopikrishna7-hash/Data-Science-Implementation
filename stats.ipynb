```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA

plt.style.use("ggplot")
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
df = pd.read_csv(list(uploaded.keys())[0])
df.head()
```

```python
data = df.drop("CustomerID", axis=1)
```

```python
numeric_data = data.select_dtypes(include=['int64', 'float64'])

plt.figure(figsize=(10,6))
sns.heatmap(numeric_data.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(numeric_data)
plt.show()
```

```python
data_encoded = pd.get_dummies(data, drop_first=True)
data_encoded.head()
```

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data_encoded)
```

```python
k_values = range(2, 11)
elbow = []
silhouette_scores = []

for k in k_values:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(scaled_data)

    elbow.append(km.inertia_)
    silhouette_scores.append(silhouette_score(scaled_data, km.labels_))
```

```python
plt.figure(figsize=(8,4))
plt.plot(k_values, elbow, marker='o')
plt.title("Elbow Method")
plt.xlabel("k")
plt.ylabel("Inertia")
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.plot(k_values, silhouette_scores, marker='o', color='green')
plt.title("Silhouette Score")
plt.xlabel("k")
plt.ylabel("Silhouette Score")
plt.show()
```

```python
k_optimal = 6   # choose best based on plots
kmeans = KMeans(n_clusters=k_optimal, random_state=42)
kmeans.fit(scaled_data)

df["Cluster"] = kmeans.labels_
df.head()
```

```python
pca = PCA(n_components=2)
components = pca.fit_transform(scaled_data)

df["PC1"] = components[:, 0]
df["PC2"] = components[:, 1]

plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x="PC1", y="PC2", hue="Cluster", palette="tab10")
plt.title("Customer Segmentation (PCA Plot)")
plt.show()
```

```python
# FIX: Analyze only numeric features
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

cluster_profile = df[numeric_cols].groupby(df["Cluster"]).mean()
print(cluster_profile)

plt.figure(figsize=(12,6))
sns.heatmap(cluster_profile, annot=True, cmap="YlGnBu")
plt.title("Cluster Profile Summary")
plt.show()
```
